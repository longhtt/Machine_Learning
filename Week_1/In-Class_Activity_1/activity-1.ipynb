### Exercise 1 — Input + Math + Strings

Start with the initial comment: to get personal information from the user

Write code to:

- Ask for name, age, height (cm), and weight (kg)
- Print lowercase, uppercase, and the length of the name
- Convert height to meters and print BMI = weight / height²
- Print weight divided by π
- Print age in days
- Split the name into characters and print them one per line

```python
name = input("What is your name?")
age = input("What is your age?")
height = input("What is your height (cm)?")
weight = input("What is your weight (kg)?")

print("Name lowercase: " + name.lower())
print("Name uppercase: " + name.upper())
print(f"Name length: ", len(name))

BMI = int(weight) / (int(height) * int(height))

pi = math.pi

print(f"BMI: ", BMI)

print(f"Weight / π = ", int(weight) / pi)

print(f"Age in days: ", int(age) * 365)

for c in name:
    print(c)

```

### Exercise 3 — List Processing
Prompt the user to enter 10 integers and store them in a list, then print:

- Mean (average)
- Min and Max
- Count of numbers > 50
- All odd numbers
- All numbers divisible by 3
- The list sorted (ascending and descending)
- The list reversed

```python
list = []
for i in range(10):
    num = input("Type a number")
    NUM = int(num)
    list.append(NUM)

print(f"Test, ", list)

print(f"Mean: ", sum(list) / len(list))
print(f"Min: ", min(list), ", Max: ", max(list))

counter = 0
odd = []
divisable_3 = []

for n in list:
    if n > 50:
        counter += 1
    if n % 2 != 0:
        odd.append(n)
    if n % 3 == 0:
        divisable_3.append(n)

print(f"Count of numbers > 50: ", counter)
print(f"Odd numbers: ", odd)
print(f"Numers dvisible by 3: ", divisable_3)
print(f"Sorted list (acending): ", sorted(list), ", (desending): ", sorted(list, reverse=True))
print(f"List reversed: ", list[::-1])
```

### Exercise 4 — Function Practice
Create 3 functions:

def add(a, b): return a + b

def multiply(a, b): return a * b

def power(a, b): return a ** b

Then write a main program that: 

- Takes 2 integers from the user
- Calls all 3 functions 
- Print results

```python
def add(a, b):
    return int(a) + int(b)

def multiply(a, b):
    return int(a) * int(b)

def power(a, b):
    return int(a) ** int(b)

A = input("Type a number for a")
B = input("Type a number for b")

print(f"add(a, b): ", add(A,B))
print(f"multiply(a, b): ", multiply(A,B))
print(f"power(a, b): ", power(A,B))
```

### Exercise 5 — Dictionary Storage

Create a program that asks for:

5 student names

1 score for each student

Store them in a dictionary like:

            {"Alice": 88, "Bob": 92, ...}

Then print:

- Class average
- Student with highest score
- All students who scored above 90

```python
list = {}

for i in range(5):
    name = input("Type a students name")
    score = int(input("Type a score for that student"))
    list[name] = score

average = sum(list.values()) / len(list)
print(f"Class average: ", average)

high_score_student = max(list, key=list.get)
print(f"Student with highest score: ", high_score_student, " - ", list[high_score_student])

above_90 = []
for name in list:
    if list[name] > 90:
        above_90.append(name)
print(f"Students who scored above 90: ", above_90)
```
